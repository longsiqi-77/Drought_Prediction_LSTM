```python
# %% [markdown]
# # 01. 数据探索性分析 (Data Exploration)
# 
# **项目**: Drought Prediction using LSTM  
# **目标**: 
# 1. 读取原始 CSV 数据 (NDVI 和 气象数据)。
# 2. 自动清洗列名 (处理 Date/DATE 大小写不一致问题)。
# 3. 可视化时间序列趋势 (NDVI vs 降水)。
# 4. 分析变量相关性 (Heatmap)。
# 
# **输出**: 此 Notebook 生成的图片可直接保存至 `results/figures/` 用于报告。

# %%
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import numpy as np

# 设置绘图风格 (根据个人喜好调整)
plt.style.use('seaborn-v0_8')
plt.rcParams['font.sans-serif'] = ['Arial'] # 英文报告推荐 Arial
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# %% [markdown]
# ## 1. 加载数据 (Load Data)
# 自动处理列名不一致的问题。

# %%
# 定义数据路径
DATA_PATH = '../data/raw'

# ================= 读取 NDVI 数据 =================
print("Loading NDVI Data...")
# 尝试读取 NDVI 文件
ndvi_file_path = os.path.join(DATA_PATH, 'MOD13Q1_NDVI.csv')
if not os.path.exists(ndvi_file_path):
    print(f"Error: File not found at {ndvi_file_path}")
else:
    # 1. 读取 CSV
    df_ndvi = pd.read_csv(ndvi_file_path)
    
    # 2. 自动寻找日期列 (包含 'date' 或 'Date' 的列)
    date_col = [c for c in df_ndvi.columns if 'date' in c.lower()][0]
    print(f"Found NDVI Date column: {date_col}")
    
    # 3. 自动寻找 NDVI 数值列 (排除日期列后的第一列通常是数据)
    value_col = [c for c in df_ndvi.columns if c != date_col][0]
    print(f"Found NDVI Value column: {value_col}")
    
    # 4. 规范化处理
    df_ndvi[date_col] = pd.to_datetime(df_ndvi[date_col])
    df_ndvi = df_ndvi.rename(columns={value_col: 'NDVI', date_col: 'Date'})
    df_ndvi = df_ndvi.set_index('Date').sort_index()

# ================= 读取 气象 数据 =================
print("\nLoading Weather Data...")
weather_file_path = os.path.join(DATA_PATH, 'NOAA_Weather.csv')
if not os.path.exists(weather_file_path):
    print(f"Error: File not found at {weather_file_path}")
else:
    # 1. 读取 CSV
    df_weather = pd.read_csv(weather_file_path)
    
    # 2. 自动寻找日期列
    weather_date_col = [c for c in df_weather.columns if 'date' in c.lower()][0]
    print(f"Found Weather Date column: {weather_date_col}")
    
    # 3. 规范化处理 (重命名为统一的 'Date')
    df_weather[weather_date_col] = pd.to_datetime(df_weather[weather_date_col])
    df_weather = df_weather.rename(columns={weather_date_col: 'Date'})
    df_weather = df_weather.set_index('Date').sort_index()

# 展示前几行
print("\n>>> Data Preview:")
display(df_ndvi.head(3))
display(df_weather.head(3))

# %% [markdown]
# ## 2. 数据融合 (Merge & Interpolate)
# 将 NDVI (16天数据) 与 气象数据 (日数据) 对齐。

# %%
# 创建全时间范围索引
start_date = max(df_ndvi.index.min(), df_weather.index.min())
end_date = min(df_ndvi.index.max(), df_weather.index.max())
full_idx = pd.date_range(start_date, end_date, freq='D')

# 合并 DataFrame
df_merged = pd.DataFrame(index=full_idx)
df_merged = df_merged.join(df_weather).join(df_ndvi)

# 插值处理: NDVI 是16天一次，需要线性插值填补中间空缺
df_merged['NDVI'] = df_merged['NDVI'].interpolate(method='linear')
# 气象数据如果有缺失，用前向填充
df_merged = df_merged.fillna(method='ffill').fillna(method='bfill')

print("Merged Data Shape:", df_merged.shape)
display(df_merged.head())

# %% [markdown]
# ## 3. 可视化趋势 (Time Series Plot)

# %%
fig, ax1 = plt.subplots(figsize=(14, 6))

# 绘制 NDVI (左轴)
color = 'tab:green'
ax1.set_xlabel('Date')
ax1.set_ylabel('NDVI (Vegetation Index)', color=color, fontweight='bold')
ax1.plot(df_merged.index, df_merged['NDVI'], color=color, alpha=0.8, linewidth=2, label='NDVI')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, alpha=0.3)

# 实例化右轴
ax2 = ax1.twinx()  

# 绘制 降水 (右轴) - 使用柱状图可能更直观
color = 'tab:blue'
ax2.set_ylabel('Precipitation (mm)', color=color, fontweight='bold')
# 为了不遮挡 NDVI，降水画成细的柱状图或者透明度高的线
ax2.bar(df_merged.index, df_merged['PRCP'], color=color, alpha=0.3, label='Precipitation', width=1.0)
ax2.tick_params(axis='y', labelcolor=color)

# 添加标题
plt.title('Vegetation Growth (NDVI) vs Precipitation', fontsize=14)
plt.show()

# %% [markdown]
# ## 4. 相关性分析 (Correlation Heatmap)
# 看看哪些气象因子跟 NDVI 最相关。

# %%
# 计算相关系数
corr = df_merged.corr(numeric_only=True)

# 绘制热力图
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", center=0, linewidths=0.5)
plt.title('Correlation Matrix of All Features')
plt.show()
```
